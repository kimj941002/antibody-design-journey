```python
import math

def distance(a, b):
    """두 좌표 사이의 유클리드 거리를 계산한다."""
    dx = a[0] - b[0]
    dy = a[1] - b[1]
    dz = a[2] - b[2]
    return math.sqrt(dx**2 + dy**2 + dz**2)

atom1 = (15.234, 22.118, 8.905)
atom2 = (18.112, 20.445, 10.332)

d = distance(atom1, atom2)
print(f"거리: {d:.2f} Å")
```
